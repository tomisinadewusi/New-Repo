```python
import pandas as pd
```

```python
#TO Import sales data (csv file) into the notebook

sales_data=pd.read_csv('sales_data.csv')
```

```python
len(sales_data)
```

```python
#Data cleaning
```

```python
#To drop duplicates in the sales dataset

sales_data.drop_duplicates()
```

```python
#To count unique values based on the product column of the dataset

sales_data['Product'].value_counts()
```

```python
#To count unique values based on the Segment column

sales_data['Segment'].value_counts()
```

```python
#To count the values in the Segment column as a proportion of the total

sales_data['Segment'].value_counts(normalize=True)
```

```python
#To compute the summary statistics of the numerical columns in the dataset

sales_data.describe()
```

```python
#To confirm if there are any missing values per column in the dataset

sales_data.isna().any()
```

```python
#To check the total number of missing values per column

sales_data.isna().sum()
```

```python
#Missing values threshold

#To calculate for 5% of the values in the column

threshold=(5/100) * len(sales_data)
threshold
```

```python
'''
If number of rows containing missing values <= 5% of the entire rows = DELETE ROWS

COLUMNS WITH NUMERICAL FIGURES
If number of rows containing missing values > 5% of the entire rows = 
                        Replace with the average of the numerical values in the column

COLUMNS WITH TEXT VALUES
If number of rows containing missing values > 5% of the entire rows = 
                        Replace with the most occuring value in the column
'''
```

```python
#To fill the missing values in the discount column with another value

sales_data.fillna({'Discount Band':'No Discount'},inplace=True)
```

```python
#To confirm if the missing values were changed

sales_data.head()
```

```python
#To recheck if there are missing values 

sales_data.isna().sum()
```

```python
#To eliminate the missing values per column

sales_data.dropna(inplace=True)

#inplace=True is to modify the original dataset imported into the current notebook
```

```python
#To recheck if there are missing values 

sales_data.isna().sum()
```
